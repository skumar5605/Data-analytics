```python
def even_odd(num):
    if num%2==0:
     print("even")
    else:
        print("odd")
    return num
even_odd(1025588)
```

```python
def factorial_n(n):
    fact=1
    for i in range(1,n+1):
        fact=fact*i
    return fact
factorial_n(6)
```

```python
###sum of n numbers
def sum_no(n):
    num=0
    for i in range(1,n+1):
        num=num+i
    return num
sum_no(5)
```

Map


map(function,iterable object)

```python
#area
import math
def area(r):
    return math.pi*(r**2)

radii=[1,2,3,4,5,6]

map(area,radii)
list(map(area,radii))
```

Map,Reduce,Filter 

```python
#C to F
#F=91C+32
temp = [('Mumbai',35),('Delhi',30),('pune',25),]
C_to_F=lambda data:(data[0],(9/5)*data[1]+32)
map(C_to_F,temp)
list(map(C_to_F,temp))
```

Filter

filter(function,data)

```python
import statistics
data=[1,2,3,4,5,6,77,88]
avg=statistics.mean(data)
print (avg)
```

```python
filter(lambda x:x>avg,data)
list(filter(lambda x:x>avg,data))
```

```python
name=['sam','ram','sham',0,0,0,'','namu']
filter(None,name)
list(filter(None,name))
```

Reduce

reduce(functions ,data)

```python
from functools import reduce
#multiply all no.
a=[1,2,3,4,5,6,7]
mul=lambda x,y:x*y
reduce(mul,a)
```
