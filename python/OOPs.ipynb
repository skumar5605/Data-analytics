```python
#creating class
class Rectangle:
    def __init__(self):
        self.length=10
        self.breadth=15
rect=Rectangle()
rect.breadth
```

```python
print("Length=",rect.length,"\nBreadth=",rect.breadth)
```

```python
class Rectangle:
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth

#instantiating class
rect=Rectangle(5,10)
#calling instance variable
rect.breadth
```

```python
#adding method to a class
class Rectangle:
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
        
    def calculate_area(self):
        return self.length*self.breadth

rect=Rectangle(15,20)  
rect.calculate_area()
```

```python
#inheritance
class Employee:
    def f1(self):
        print("Hii")
class Department(Employee): 
    pass

emp=Employee()
emp.f1()
```

```python
dep=Department()
dep.f1()
```

```python
#OVERRIDING
class shape:
    def set_color(self,color):
        self.color=color

    def calculate_area(self):
        pass

    def color_the_shape(self):
        color_price={'red':10,'blue':15,'green':19}
        return self.calculate_area()*color_price[self.color]

class Circle(shape):
    pi=3.14
    def __init__(self,radius):
        self.radius=radius

    def calculate_area(self):
        return Circle.pi*self.radius*self.radius

c=Circle(5)

#accesing set_color from parent class
c.set_color('red')
print("circle with radius=",c.radius,'\nwhen colored',c.color,'\ncosts=',c.color_the_shape())
```
